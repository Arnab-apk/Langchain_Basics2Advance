# Messages in LangChain

LangChain uses message objects to represent a conversation history passed to chat models.

## Common Message Types

- **SystemMessage**: Sets behavior and rules for the assistant.
- **HumanMessage**: Represents user input.
- **AIMessage**: Represents model responses.
- **ToolMessage**: Carries tool output back into the conversation.

## Why Messages Matter

Messages preserve context across turns so the model can reason over prior instructions, user questions, and tool results.

## Minimal Example

```python
from langchain_core.messages import SystemMessage, HumanMessage

messages = [
    SystemMessage(content="You are a helpful coding assistant."),
    HumanMessage(content="Explain what an API is in one sentence."),
]
```
